```python
host_1 = '127.0.0.1'
host_2 = '192.168.1.90'
port_1 = 7496
port_2 = 7497
port_3 = 4002
ib = IB()
ib.connect(host_1, port_2, clientId=47)
```

```python
contract = Stock('TSLA', 'SMART', 'USD')
ib.qualifyContracts(contract)
```

```python
load_from_csv = True
if load_from_csv:
    bars_df = pd.read_csv('TSLA_1min.csv', parse_dates=['date'])
else:
    endDateTime = (datetime.datetime.now() + datetime.timedelta(days=-1)).strftime('%Y%m%d 21:00:00 US/Eastern')
    # endDateTime = '20241109 16:20:00 US/Eastern'
    keepUpToDate = False
    useRTH = False
    bars = ib.reqHistoricalData(
        contract,
        endDateTime='',
        durationStr='1 D',
        barSizeSetting='1 min',
        whatToShow='TRADES',
        useRTH=useRTH,
        keepUpToDate=keepUpToDate,
        formatDate=1)
    bars_df = util.df(bars)
```

```python
import seaborn as sns

# Create a pivot table for the heatmap
heatmap_data = bars_df.pivot_table(index=bars_df.index, columns='date', values='close')

# Plot the heatmap
plt.figure(figsize=(12, 6))
sns.heatmap(heatmap_data, cmap='viridis')
plt.xlabel('Date')
plt.ylabel('Index')
plt.title('Heatmap of Close Prices')
plt.show()
```
