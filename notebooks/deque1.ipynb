```python
def slice_deque(dq, start=None, end=None):
    length = len(dq)
    if length == 0:
        return collections.deque()
    
    if start is None:
        start = 0
    elif start < 0:
        start += length
    
    if end is None:
        end = length
        end += 1
    elif end < 0:
        end += length
        end += 1
    
    start = max(0, min(start, length))
    end = max(0, min(end, length))
    
    return collections.deque(itertools.islice(dq, start, end))
```

```python
# Example usage
sliced_dq = slice_deque(dq, -400, -10)
print(sliced_dq)
```

```python
%timeit sliced_dq = slice_deque(dq, -4000, -1000)
```

```python
def slice_deque(dq, start, end):
    dq.rotate(-start)
    start, end = 0, end - start
    dq.rotate(start)
    if start < 0:
        start += len(dq)
    if end < 0:
        end += len(dq)
    return collections.deque(list(dq)[start:end])

# Example usage
sliced_dq = slice_deque(dq, -4, )
print(sliced_dq)
```
